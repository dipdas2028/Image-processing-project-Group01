This notebook demonstrates the complete classical Image Processing pipeline for detecting document boundaries, rectifying perspective distortion via a $3 \times 3$ projective homography, and performing illumination-invariant adaptive binarization.

```python
import sys
sys.path.append('..')
import cv2
import matplotlib.pyplot as plt
from src.dataset_generator import generate_benchmark_dataset
from src.scanner import scan_document
from src.evaluate import run_evaluation
```

```python
# Generate benchmark images and inspect Scenario 2 (Shadow + Noise)
dataset = generate_benchmark_dataset('../data/sample_inputs')
sample = dataset[1]['image']
result = scan_document(sample, out_size=(600, 800))
print(f"Processed in {result['processing_time_ms']:.2f} ms")
print('Detected Corners:\n', result['corners'])
```

```python
# Visualize Input, Edges, Warped Color, and Final Binary Scan
fig, axes = plt.subplots(1, 4, figsize=(16, 5))
axes[0].imshow(cv2.cvtColor(sample, cv2.COLOR_BGR2RGB)); axes[0].set_title('Input Capture')
axes[1].imshow(result['closed_edges'], cmap='gray'); axes[1].set_title('Closed Edges')
axes[2].imshow(cv2.cvtColor(result['warped_color'], cv2.COLOR_BGR2RGB)); axes[2].set_title('Perspective Warp')
axes[3].imshow(result['scanned_binary'], cmap='gray'); axes[3].set_title('Final Binary Scan')
for ax in axes: ax.axis('off')
plt.tight_layout()
plt.show()
```
